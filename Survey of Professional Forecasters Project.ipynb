```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy.stats import uniform
```

```python
#true distribution

true_mean = 3.3
true_std = 1.4
n = 10000

data = np.random.normal(true_mean,true_std,n)
```

```python
def cdf(data,mean,std):
    return norm.cdf(data,mean,std)
```

```python
alice = cdf(data,3.3,1.4)
bob = cdf(data,5.3,1.4)
chris = cdf(data,3.3,0.6)
dan = cdf(data,5.3,0.6)
#extra
jon = uniform.cdf(data) #uniform cdf
```

```python
fig,ax = plt.subplots(5,1,figsize=(10,10))

ax[0].hist(alice,bins=20,edgecolor='black'); 
ax[0].set_title("Alice");
ax[0].grid(alpha=0.4, linestyle='--')
ax[0].set_ylabel("Count")
# Alice's graph matches true distribution therefore all the cdf values are uniformly between 0 & 1

ax[1].hist(bob,bins=20,edgecolor='black');
ax[1].set_title("Bob");
ax[1].grid(alpha=0.4, linestyle='--')
ax[1].set_ylabel("Count")
#In Bob's graph he thinks the mean is higher than what it actually is so the data looks too low for him therefore his CDF values are mostly less than 0.5

ax[2].hist(chris,bins=20,edgecolor='black');
ax[2].set_title("Chris");
ax[2].grid(alpha=0.4, linestyle='--')
ax[2].set_ylabel("Count")
#In Chris's graph he thinks standard deviation is smaller, he thinks the values are closer to the mean however the real data is more spread out thus making his values seem really high or really low. this causes the values to cluster near the end points 

ax[3].hist(dan,bins=20,edgecolor='black');
ax[3].set_title("Dan");
ax[3].grid(alpha=0.4, linestyle='--')
ax[3].set_ylabel("Count")
# In Dan's graph both his mean and std are wrong, his mean is too far right and the std is too low, so when he plugs in the real data they all seem unlikely and low this is shown in the CDF values clustering near 0

ax[4].hist(jon,bins=20,edgecolor='black');
ax[4].set_title("Jon");
ax[4].grid(alpha=0.4, linestyle='--')
ax[4].set_ylabel("Count")
#Jon thinks the data come from a uniform distribution so the values should be always between 0 and 1. But the normal distribution is centered aroudn 3.3 so most valyes are greater than 1. This leads to the CDF values clustering at 1

ax[-1].set_xlabel("CDF value")
plt.tight_layout()
```
