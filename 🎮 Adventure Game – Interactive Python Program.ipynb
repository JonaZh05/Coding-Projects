```python
#Everything Together
#modules
import random
import time

#Dice Class
class Dice:
  def dice_20():
    return random.randint(1,20)
  def dice_6():
    return random.randint(1,6)

class Monster:
  def __init__(self, name, hp, attack):
    self.name = name
    self.health_points = hp
    self.attack = attack

  def show_stats(self):
    print(f"{self.name}: has {self.health_points} HP, {self.attack} Attack")

class Player:
  def __init__(self, name):
    self.name = name
    self.gold = 0
    self.inventory = []

    self.HealthPoints = Dice.dice_20()
    self.Attack = Dice.dice_6()
    self.Defense = Dice.dice_6()
    self.weapon_damage = 0

  #Weapon
  def call_weapon(self):
    print("Rolling weapon...")
    time.sleep(2)
    weapon_roll = Dice.dice_6()
    if weapon_roll in [1, 2]:
      self.weapon_damage = 1
      print(f"The weapon you have rolled are your fists")
      print("\n")
    elif weapon_roll in [3, 4]:
      self.weapon_damage = 2
      print(f"The weapon you have rolled is a stick")
      print("\n")
    else:
      self.weapon_damage = 3
      print(f"The weapon you have rolled is a sword")
      print("\n")
    #total attack
    self.total_damage = self.Attack + self.weapon_damage
    time.sleep(1)
    print("Rolling character stats...")
    time.sleep(1)
    #add a pause here

  def show_status(self):
    print(f"{self.name}'s stats are:")
    print(f"Health Points - {self.HealthPoints}")
    print(f"Base Attack - {self.Attack}")
    print(f"Weapon Attack - {self.weapon_damage}")
    print(f"Total Attack Damage - {self.total_damage}")
    print(f"Defense - {self.Defense}")
    print(f"Gold - {self.gold}")
    time.sleep(2)

class Game:
  def __init__(self, player_name):
    self.player = Player(player_name)
    self.shop = Shop()
    self.wave = 0
    self.monsters = [
    Monster("Slime", 10, 3),
    Monster("Goblin", 12, 5),
    Monster("Witch", 15, 7),
    Monster("Dragon", 25, 9)] #Final boss

  def display_instructions(self):
    print("\n")
    print("Welcome to the Adventure Game: ")
    print("You will face 3 waves of enemies and a final boss battle")
    print("All actions taken in the game will be decided by dice roll")
    print("Luck and strategy will determine your fate.")
    time.sleep(3)

    print("\n")

    print("Character Creation:")
    print("A dice will be rolled to determine the Health, Attack, and Defense of your character")
    print("A dice will also be rolled to determine the weapon that you will be utilizing throughout your journey")
    time.sleep(3)

    print("\n")

    print("Gameplay Overview:")
    print("In each wave, you will face a different monster")
    print("You can choose to ATTACK or DEFEND")
    print("At the end of every wave the monster will drop gold that you can utilize to purchase upgrades at the shop")
    print("Survive all 3 monster waves to face the FINAL BOSS")
    time.sleep(3)

    print("\n")

    print("The game ends when:")
    print("Your health reaches 0 or when you beat the final boss!")
    print("May RNG be in your favor!")
    time.sleep(3)
    print("\n")

  def start_game(self):
    self.player.call_weapon()
    time.sleep(2)
    while self.wave < 4: #3 waves and a boss
      self.player.show_status()
      time.sleep(1)
      print("\n")
      monster = self.monsters[self.wave] #sets monster as a variable and then from the list of monsters it uses self.wave to keep track of which monster you want based on the wave you are on
      monster.show_stats()

      self.battle(monster)
      if self.wave == 0: #First wave
        monster = self.monsters[0] #monster should be Slime
        self.player.gold += 3
        print("You earned 3 gold for beating wave 1")
      if self.wave == 1: #Second wave
        monster = self.monsters[1] #monster should be Goblin
        self.player.gold += 4
        print("You earned 4 gold for beating wave 2")
      if self.wave == 2: #Third wave
        monster = self.monsters[2] #monster should be Witch
        self.player.gold += 5
        print("You earned 5 gold for beating wave 3")
      if self.wave == 3: #Boss wave
        monster = self.monsters[3] #monster should be Dragon
        print("Congratulations, you have defeated the final boss! Here is a participation award")
        break

      self.shop.buy_upgrade(self.player)
      self.wave += 1


  def battle(self, monster):
    print(f"Wave {self.wave +1}: You are fighting the {monster.name}!")
    while monster.health_points > 0 and self.player.HealthPoints > 0:
    #Battle only when both have > 0 Hp
      action = input("What action do you want to take? (attack or defend) ").lower()

      if action == "attack":
        print("You have chosen to attack")
        print("\n")
        time.sleep(2)
        player_damage = self.player.total_damage
        monster.health_points -= player_damage
        print(f"You attacked the {monster.name} and dealt {player_damage} damage! {monster.name} now has {monster.health_points} Hp.")

      elif action == "defend":
        defense_roll = Dice.dice_20()
        print(f"You are defending with a 20% chance to counter damage")
        time.sleep(2)
        if defense_roll <= 4: #20% chance
          time.sleep(2)
          counter_damage = self.player.total_damage//2 #half of total_damage as counter damage
          monster.health_points -= counter_damage
          time.sleep(1)
          print(f"You have parried {monster.name}'s attack and dealt {counter_damage} damage back! {monster.name} now has {monster.health_points} Hp.")
          time.sleep(1)
        else:
          time.sleep(1)
          print("You defended but didn't counter.")
          time.sleep(2)

      else: #edgecase
        print("Invalid action. Please enter (attack or defend). ")
        continue

      if monster.health_points <= 0:
        print(f"You have defeated the {monster.name}!")
        time.sleep(2)
        print("\n")
        break #exit combat

      monster_damage = monster.attack #monster damage
      damage_taken = max(monster_damage - self.player.Defense, 0) #Actual damage after def and counting for negative damage being taken
#googled: https://stackoverflow.com/questions/37398903/how-to-avoid-negative-damage-is-healing-the-enemy
#checked max function: https://www.w3schools.com/python/ref_func_max.asp
      self.player.HealthPoints -= damage_taken #subtract from current hp
      print("\n")
      print(f"The {monster.name} attacks dealing {monster.attack} damage to you! Your defense of {self.player.Defense} made you take {damage_taken} damage.")
      time.sleep(2)

      if self.player.HealthPoints <= 0:
        print(f"You have lost to {monster.name}! Game over.")
        time.sleep(1)
        print("Try being luckier next game.")



#Shop
class Shop:
  def __init__(self):
    self.upgrades = {"HP Upgrade": 3,
                  "Defense Upgrade": 3,
                  "Attack Upgrade": 3,
                  "Weapon Upgrade": 3}

  #Code adapted from JiaXi_Yu's lab_9
  def buy_upgrade(self, player): #Buying upgrades
    while player.gold >= 3:
      time.sleep(2)
      print("Welcome to the Shop!")
      time.sleep(2)
      print("Here are the items available:")
      for item in self.upgrades:
          cost = self.upgrades[item]
          print(f"{item}: {cost} gold")
      print("A dice will be rolled to determine the value of your upgrade. (+1/+2/+3)")
      print("\n")
      time.sleep(2)
      print("Here are your stats:")
      player.show_status()

      upgrade_choice = input("Input the upgrade you would like to buy(Copy & paste the upgrade name) or enter LEAVE(in caps) to exit shop ")
      print("\n")
      time.sleep(2)
      if upgrade_choice == "LEAVE":
        print("You have left the shop")
        break #Exits Shop

      if upgrade_choice in self.upgrades:
        upgrade_cost = self.upgrades[upgrade_choice]
        if player.gold >= upgrade_cost:
          player.gold -= upgrade_cost
        print(f"You have bought {upgrade_choice} for 3 gold, you have {player.gold} gold.")

        #roll upgrade boost
        upgrade_boost = Dice.dice_6()
        if upgrade_boost in [1, 2]:
        #upgraded_stat is actual amount being added to stat
          upgraded_stat = 1
        elif upgrade_boost in [3, 4]:
          upgraded_stat = 2
        else:                #[5,6]
          upgraded_stat = 3

          #incorporating shop upgrades
        time.sleep(2)
        if upgrade_choice == "HP Upgrade":
          player.HealthPoints += upgraded_stat
          print(f"Your HP increased by {upgraded_stat}!")
        elif upgrade_choice == "Defense Upgrade":
          player.Defense += upgraded_stat
          print(f"Your Defense increased by {upgraded_stat}!")
        elif upgrade_choice == "Attack Upgrade":
          player.Attack += upgraded_stat
          player.total_damage = player.Attack + player.weapon_damage
          print(f"Your Base Attack increased by {upgraded_stat}!")
        elif upgrade_choice == "Weapon Upgrade":
          player.weapon_damage += upgraded_stat
          player.total_damage = player.Attack + player.weapon_damage
          print(f"Your Weapon Damage increased by {upgraded_stat}!")
          time.sleep(2)
      else:
        print(f"{upgrade_choice} does not exist, please enter a valid upgrade")

    else:
      print("You don’t have enough gold for an upgrade.")
      print("Leaving the shop.")
      print("\n")
      time.sleep(2)

#Play Again Function
def play_again():
    while True:
      print("\n")
      response = input("Do you want to play again? (yes/no): ").lower()
      if response == "yes":
        return True
      elif response == "no":
        return False
      else:
        print("Invalid, Please enter (yes/no): ")
```

```python
def main():
  while True:
    player_name = input("What is your name Adventurer? ")
    time.sleep(1)
    game = Game(player_name)
    game.display_instructions()
    game.start_game()
    if not play_again(): #if play_again is false
      print(" Thanks for playing! Goodbye. ")
      return
main()
```
